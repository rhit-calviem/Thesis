# Image Crop Tool
Adjust the crop region in **Cell 1** until it looks right, then run **Cell 2** to save.

```python
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# ── CONFIG ──────────────────────────────────────────────────────────────────
IMAGE_PATH = "building_gan.png"

# Crop box in pixels: (left, upper, right, lower)
# Tip: upper-left corner is (0, 0)
LEFT   = 200
UPPER  = 0
RIGHT  = 380
LOWER  = 160
# ────────────────────────────────────────────────────────────────────────────

img  = Image.open(IMAGE_PATH)
crop = img.crop((LEFT, UPPER, RIGHT, LOWER))

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Left: full image with crop rectangle
axes[0].imshow(img)
rect = patches.Rectangle(
    (LEFT, UPPER), RIGHT - LEFT, LOWER - UPPER,
    linewidth=2, edgecolor='red', facecolor='none'
)
axes[0].add_patch(rect)
axes[0].set_title(f"Full image  ({img.width} × {img.height} px)")
axes[0].axis('off')

# Right: cropped region
axes[1].imshow(crop)
axes[1].set_title(
    f"Crop preview  ({crop.width} × {crop.height} px)\n"
    f"({LEFT}, {UPPER}) → ({RIGHT}, {LOWER})"
)
axes[1].axis('off')

plt.tight_layout()
plt.show()
print("Happy with the crop? Run Cell 2 to save it.")
```

```python
# ── SAVE ─────────────────────────────────────────────────────────────────────
OUTPUT_PATH = "cropped_output_gan.jpg"   # <-- change if needed
# ─────────────────────────────────────────────────────────────────────────────

crop.save(OUTPUT_PATH)
print(f"Saved: {OUTPUT_PATH}  ({crop.width} × {crop.height} px)")
```
